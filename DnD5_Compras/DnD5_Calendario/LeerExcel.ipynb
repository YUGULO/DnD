```python
import gspread
from oauth2client.service_account import ServiceAccountCredentials

# Define el alcance (permisos) que necesitas
scope = ['https://spreadsheets.google.com/feeds', 'https://www.googleapis.com/auth/drive']

# Carga las credenciales desde el archivo JSON
creds = ServiceAccountCredentials.from_json_keyfile_name('credentials.json', scope)

# Autoriza el cliente de Google Sheets
client = gspread.authorize(creds)

# Abre la hoja de cálculo por su URL
sheet = client.open_by_url('URL_DE_TU_HOJA_DE_CALCULO').sheet1

# Obtiene el valor de la celda A1
celda_a1 = sheet.acell('A1').value

# Imprime el valor en pantalla
print(celda_a1)
```
